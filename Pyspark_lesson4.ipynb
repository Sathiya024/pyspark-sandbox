```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName("DataFrame").getOrCreate()
```

```python
spark
```

```python
df_pyspark = spark.read.csv('test1.csv',header=True,inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```
